### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Thing

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
def anagram_checker(word_a, word_b):
    """ 
    This function checks if two words are anagrams of each other and returns a boolean value. 
    The function is not case-sensitive and ignores capitalization of letters in the input string. 
    Anagram definition: A word or phrase formed by rearranging the letters of a different word or 
    phrase, typically using all the original letters exactly once. 
    Example: "listen" is an anagram of "silent". 
    """

    # Step 1: Converting both variables to string data type and lowercase format
    word_a = str(word_a).lower()
    word_b = str(word_b).lower()

    # Step 2: Using an "equal" operator and Sort function to check if the two words contain the same letters and thus are considered anagrams. 
    return sorted(word_a) == sorted(word_b)

# Running code to test the function - Test A
anagram_checker("Silent", "listen")
````

````python
# Testing the function - Test B
anagram_checker("Silent", "Night")
````

````python
# Testing the function - Test C
anagram_checker("night", "Thing")
````

### Part 2: Expanding the functionality of the Anagram Checker

Using your existing and functional anagram checker, let's add a boolean option called `is_case_sensitive`, which will return `True` or `False` based on if the two compared words are anagrams and if we are checking for case sensitivity.

````python
def anagram_checker(word_a, word_b, is_case_sensitive: bool = False):
    """ 
    This function checks if two words are anagrams of each other and returns a boolean value. 
    The function is case-sensitive and returns 'True' value only if two words are anagrams AND 
    have the exact same lettercase formatting. Otherwise, the function returns 'False'. 
    Anagram definition: A word or phrase formed by rearranging the letters of a different word 
    or phrase, typically using all the original letters exactly once. 
    Example: "listen" is an anagram of "silent". 
    """

    # Step 1: Removing data type conversion to lowercase format as seen in Part 1
    #word_a = str(word_a).lower()
    #word_b = str(word_b).lower()

    # Step 2: Checking for case sensitivity with an output of a boolean value True/False
    if is_case_sensitive:
        return sorted(word_a) == sorted(word_b)
    else:
        # Step 3: Checking if two words are considered anagrams when sorted and converted to lowercase data type when is_case_sensitive output returns False.
        return sorted(word_a.lower()) == sorted(word_b.lower())

# Running code to test the function - Test A:
anagram_checker("Silent", "listen", False) # True
````

````python
# Running code to test the function - Test B:
anagram_checker("Silent", "listen", True) # False
````

````python
# Running code to test the function - Test C:
anagram_checker("Silent", "Listen", True) # False
````
